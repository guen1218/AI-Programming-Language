## 학습 되돌아보기 (2)

```python
import random
import math
ban = round(random.random()*10, 2)
pi = round(math.pi, 4)

print("원의 반지름: ", ban)
print("원주율 pi: ", pi)
print(f"반지름 {ban}인 원의 면적은 {round(ban * ban * pi, 2)}")
```

## 학습 되돌아 보기(3)

```python
import math as m
import statistics as s
st = [80, 99, 77, 65, 92, 74, 82]
if __name__ == '__main__':
    print("1! = ",m.factorial(1))
    print("6! = ",m.factorial(6))
    print("11! = ",m.factorial(11))
    print("16! = ",m.factorial(16))
    print("\n")
    print(st)
    print("중앙 값: ",s.median(st))
    print("평균: ",round(s.mean(st),2))
    print(f"분산: {s.variance(st):.2f}")
    print("표준편차: ",round(s.stdev(st),2))
```
