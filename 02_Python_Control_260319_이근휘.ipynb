```python
#while문 연습 : 입력값에 따라 반복 종료하기
#사용자로부터 숫자를 하나씩 입력 받아서 총합과 평균을 출력
#사용자가 0을 입력하면 숫자 입력을 멈춤

sum = count = 0
num = int(input("숫자 입력 (0 : 끝) : "))
while num != 0:
    sum += num
    count += 1
    num = int(input("숫자 입력 (0 : 끝) : "))

#0으로 나누기를 하면 ZeroDivisionError 발생
if count != 0:
    average = sum / count 
else :
    average = 0
    
print (f">>총합 : {sum}, 평균 : {average:.2f}")
```

```python
#무한 루프와 분기문 break 활용

sum = count = 0

while True :
    number = int(input("숫자 입력 (0 : 끝) : "))
    if number == 0:
        break
    sum += number
    count += 1
    
print (f">>합 : {sum}, 평균 : {sum/count:.2f}")
```

```python
#숫자를 입력 받아서 1부터 해당 숫자까지 더하여 출력
num = int(input("숫자를 입력하세요 : "))

sum = 0
for _ in range (1,num+1) :
    # print(_)
    sum += _

print (f"1부터 {num}까지의 합은 {sum}입니다.")
```

```python
#입력 받은 숫자에 해당하는 구구단 출력
num = int(input("숫자를 입력하세요 : "))

print (f"[구구단 {num}단]")
for _ in range (1, 10):
    print(f"{num} X {_} = {num*_}")
```

```python
#입력받은 숫자의 팩토리얼(Factorial) 값을 출력
num = int(input("숫자를 입력하세요 : "))
fac = 1;

for _ in range (num, 0, -1):
    fac *= _
print (f"{num}! = {fac}")
```
