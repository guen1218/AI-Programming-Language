## 튜플 (tuple)의 개념

<p> 특징</p>
<p>- 항목 간에 순서가 있음 (ordered) → 항목별 위치가 정해져 있음 (indexing)</p>
<p>- 항목 자료 값은 중복 되도 무관</p>
<p>- 수정 불가능 (immutable) → 항목을 추가, 삭제, 수정하려고 하면 오류 발생</p>
<p>- 항목은 정수, 실수, 문자열, 리스트 등이 모두 가능</p>
<p>• 하나의 리스트에 다양한 데이터 타입을 담을 수 있음</p>

<p>t1=(1,2,3) #튜플 = (소괄호)</p>
<p>t2=('a', 'b', ('ab','cd'))</p>
<p>t3=('tuple',(1,2),[3,4]) # 튜플 안 튜플, 리스트 가능</p>
