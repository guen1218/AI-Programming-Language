## 튜플 (tuple)의 개념

<p> 특징</p>
<p>- 항목 간에 순서가 있음 (ordered) → 항목별 위치가 정해져 있음 (indexing)</p>
<p>- 항목 자료 값은 중복 되도 무관</p>
<p>- 수정 불가능 (immutable) → 항목을 추가, 삭제, 수정하려고 하면 오류 발생</p>
<p>- 항목은 정수, 실수, 문자열, 리스트 등이 모두 가능</p>
<p>• 하나의 리스트에 다양한 데이터 타입을 담을 수 있음</p>

<p>t1=(1,2,3) #튜플 = (소괄호)</p>
<p>t2=('a', 'b', ('ab','cd'))</p>
<p>t3=('tuple',(1,2),[3,4]) # 튜플 안 튜플, 리스트 가능</p>

```python
#지우고 싶은 인덱스 값을 받아 지우는 문제
test_list =(11, 23, 31, 'arirang', 'bom', '0', 10, 11)
print(test_list)
b='네'
while b == '네' and test_list:
    a = int(input("지우고 싶은 곳 인덱스 입력"))
    if a <= len(test_list):
        test_list.pop(a)
    else :
        print("다시 선택하세요")
    print(test_list)
    b = input("더 지우시겠습니까? 네/아니오")
```

```python
#숫자를 입력받아 배열을 정렬하는 문제
a_list=('CCC', 'B', 'DD', 'A')
print("1. 알파벳 정렬, 2. 알파벳 역순, 3. 길이 정렬, 4. 반대로 정렬")
num = input()

if num == "1":
    b_list = sorted(a_list)
elif num =="2":
    b_list = sorted(a_list, reverse=True)
elif num =="3":
    b_list = sorted(a_list, key=len)
elif num=="4":
    b_list = list(reversed(a_list))

print('정렬 후 데이터 ',b_list)
print('원본 데이터 : ',a_list)

#리스트를 새로 만들어서 그런지 정렬이 잘 되네요
```

```python
#리스트 안에 어떤 숫자가 어디 있고 몇개 있는지 알려주는 문제
test_list =(1, 5, 3, 4, 1, 3, 1, 4, 5, 6, 3, 4, 6,7, 8, 9, 0, 9,8,7,9,0,6,7,5,1)

print("리스트 : ",test_list)
num=int(input("찾을 숫자 입력 : "))

print("찾는 숫자의 여부 : ",num in test_list)
if num in test_list:
    print("찾는 숫자가 처음 위치한 곳 : ", test_list.index(num))
    print("찾는 숫자의 개수: ", test_list.count(num))
```
