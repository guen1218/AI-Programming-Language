```python
prices = {'아메리카노':2000, '라떼':3000, '아메리카노':3500, '녹차':2500}
print(prices)
```

```python
# [실습] 앞 페이지에서 생성한 커피 메뉴로 실습해 보기
it = prices.items()
key = prices.keys()
val = prices.values()
print(it)
print(key)
print(val)
for i in prices:
    print(i)
```

```python
# [실습] 앞 페이지에서 생성한 커피 메뉴로 실습해 보기
print(prices)
data = prices.pop('라떼')
print(prices)
del prices['아메리카노']
print(prices)
prices.clear()
print(prices)
```

```python
def mos(a):
    dic ={
        'a': '·-', 'b': '-···', 'c': '-·-·', 'd': '-··' , 'e': '·', 'f': '··-·', 'g': '--·', 
        'h': '····' , 'i': '··', 'j': '·---', 'k': '-·-', 'l': '·-··', 'm':'--' , 'n': '-·', 
        'o': '---' , 'p': '·--·', 'q': '--·-', 'r': '·-·', 's':'···' , 't': '-', 'u': '··-', 
        'v':'···-' , 'w': '·--', 'x':'-··-' , 'y':'-·--' , 'z': '--··', 
    }
    return dic[a]

a = list(input("> 알파벳 문자열 입력 : ").lower())
print(">> 모스부호 : ",end='')
for i in a :
    print(mos(i),end=' ')
```
