## 문자열을 원하는 형식으로 만들기

<p>string ="ABC"</p>
<p>print(f"|{string:10}|") # 10칸쓸게요 본인포함</p>
<p>print(f"|{string:>10}|") # 오른쪽정렬</p>
<p>print(f"|{string:-^10}|") # 가운데 정렬 사이에 - 채우기</p>
<p>print(f"|{string:*>10}|") # 오른쪽 정렬 사이에 * 채우기</p>
<hr>
<p>number = 1234</p>
<p>print(f"|{number:10}|")</p>
<p>print(f"|{number:<10}|")</p>
<p>print(f"|{number:-^10}|")</p>
<p>print(f"|{number:*>10}|")</p>
<p>print(f"|{number:010}|") # 0 채우기</p>

```python
number = int(input("숫자 입력"))
print(f"|{number:15}|")
print(f"|{number:<15}|")
print(f"|{number:^15}|")
print(f"|{number:!^15}|")
print(f"|{number:%>15}|")
print(f"|{number:>15}|")
print(f"|{number:015}|")
```

### 문자열 구성 확인하기

print('1234kg'.isdigit()) # 숫자로만 구성돼있는가

print('python'.isalpha()) # 알파벳으로만 구성돼있는가

print('pass123'.isalnum()) # 알파벳/숫자로 구성돼있는가

print('python'.islower()) # 소문자로 구성돼있는가

print('Python'.isupper()) # 대문자로 구성돼있는가

print(' '.isspace()) # 공백으로 구성돼있는가

```python
a = input("입력")

if a.isdigit(): print("숫자")
if a.isalpha(): print("알파벳")
if a.isalnum(): print("숫자알파벳")
if a.islower(): print("소문자")
if a.isupper(): print("대문자")
if a.isspace(): print("공백")
```

### [실습] 알파벳 두 개를 입력 받아서 두 개 사이에 몇 개의 알파벳이 있는지 계산

```python
a,b=input().lower().split()
if a == b:
    print(0)
elif ord(a) > ord(b):
    print(ord(a)-ord(b)-1)
else :
    print(ord(b)-ord(a)-1)
```

### [실습] 모스 부호 번역기 영어로 된 문장을 모스 부호로 변환하기

```python
a = list(input("> 알파벳 문자열 입력 : ").upper())
rsu=''
def fun(a):
    if a == "A":
        return ".-"
    elif a == "B":
        return "-..."
    elif a == "C":
        return "-.-."
    elif a == "D": 
        return "-.."
    elif a == "E":
        return "."
    elif a == "F":
        return "..-."
    elif a == "G":
        return "--."
    elif a == "H":
        return "...."
    elif a == "I":
        return '..'
    elif a == "J":
        return '.---'
    elif a == "K":
        return "-.-"
    elif a == "L":
        return '.-..'
    elif a == "M":
        return '--'
    elif a == "N":
        return "-."
    elif a == "O":
        return '---'
    elif a == "P":
        return '.--.'
    elif a == "Q":
        return '--.-'
    elif a == "R":
        return ".-."
    elif a == "S":
        return '...'
    elif a == "T":
        return '-'
    elif a == "U":
        return '..-'
    elif a == "V":
        return "...-"
    elif a == "W":
        return '.--'
    elif a == "X":
        return '-..-'
    elif a == "Y": 
        return '-.--'
    elif a == "Z":
        return "--.."
    
for i in a:
    rsu = rsu + fun(i) + " "
print(">> 모스 부호 : ",rsu)
```
