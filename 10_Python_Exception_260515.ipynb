```python
try:
    num_list = [2, 6, 34, 65, 11]
    index = int(input("인덱스입력 : "))
    print(num_list[index])
except (ValueError, IndexError):
    print(f"0부터 {len(num_list)-1}까지의 정수를 입력하세요.")
except Exception as e:
    print(f"예외가 발생했습니다 : {e} -> {type(e)}")
```

```python
while True:
    try:
        number = int(input("양의 정수 입력 :"))
        if number <= 0 :
            raise ValueError()
    except ValueError:
        print("양의 정수를 입력하세요.")
    else:
        sum = 0 
        for i in range(1, number+1):
            sum += i
        print(f">> 1부터 {number}까지의 합 : {sum}")
        break
```

```python
uyear = {1:'freshman', 2:'sophomore', 3:'junior', 4:"senior"}

uinput = int(input("대학교 몇 학년이지요? "))

try:
    if uinput not in uyear.keys() :
        raise Exception("1 ~ 4 사이 정수를 입력하세요")
except Exception as e:
    print(f"예외 발생 이름: {type(e)}")
    print("예외 발생 이유: ",e)
else:
        print(f"{uinput}학년: {uyear[uinput]}")    
finally:
    print("예외 처리가 잘되는군요!")
```
