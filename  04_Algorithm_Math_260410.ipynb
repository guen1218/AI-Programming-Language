```python
def prime_number(N):
    is_prime_number = True
    if N <= 1 or N % 2 == 0:
        is_prime_number = False
    else :
        for j in range (3, N//2):
            if N%j == 0:
                is_prime_number = False
                break
    return is_prime_number
```
